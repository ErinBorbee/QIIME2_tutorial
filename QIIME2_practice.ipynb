### 3. Sequence trimming
Once your reads are imported, you can use the cutadapt plugin within QIIME2 to remove poor quality bases from the front and back ends of your reads. Usually, we would also make sure to remove primer sequences but in this case the sequencing center has already done that. When deciding how many bases to trim off, make sure you maintain at least 10bp of overlap. You can use your FastQC output to determine how long the sequences are to start and keep in mind that after merging the 16S V4 region should be ~290bp. In some cases this may mean you cannot trim any bases from your sequences, but use your best judgement.

https://docs.qiime2.org/2024.10/plugins/available/cutadapt/index.html

### 7. Differential abundance
This last step will use analysis of composition (ANCOM) to look at differential abundance of different taxa/sequences across samples from different groups in the metadata. Test out different variable to see if there are any potential effects. Since we are dealing with a small subset of the data we may not have the statistical power to detect significance.
